```python
def convertir_temperatura(valor, desde, hacia):
    """
    Convierte temperaturas entre Celsius, Fahrenheit y Kelvin.
    
    Parámetros:
        valor (float): temperatura a convertir
        desde (str): escala de origen ('C', 'F' o 'K')
        hacia (str): escala de destino ('C', 'F' o 'K')
    
    Retorna:
        float: temperatura convertida
    """
    desde, hacia = desde.upper(), hacia.upper()
    if desde == hacia:
        return valor

    # Convertir primero a Celsius
    if desde == 'F':
        celsius = (valor - 32) * 5/9
    elif desde == 'K':
        celsius = valor - 273.15
    else:
        celsius = valor

    # Convertir de Celsius a la escala destino
    if hacia == 'F':
        return celsius * 9/5 + 32
    elif hacia == 'K':
        return celsius + 273.15
    else:
        return celsius


# Ejemplos de uso
print(convertir_temperatura(25, 'C', 'F'))  # 77.0
print(convertir_temperatura(77, 'F', 'K'))  # 298.15
print(convertir_temperatura(300, 'K', 'C')) # 26.85
temperatura='s'
while temperatura=='s':
    try:
        temperatura = float(input("Ingrese la temperatura a convertir (o 'exit' para salir): "))
        desde = input("Ingrese la escala de origen (C, F, K): ")
        hacia = input("Ingrese la escala de destino (C, F, K): ")
        resultado = convertir_temperatura(temperatura, desde, hacia)
        print(f"Temperatura convertida: {resultado:.2f} {hacia.upper()}")
    except ValueError:
        print("Entrada no válida. Por favor, intente de nuevo.")
    except KeyboardInterrupt:
        print("\nConversión terminada.")
        break
    temperatura=input("¿Desea realizar otra conversión? (s/n): ").strip().lower()
```
