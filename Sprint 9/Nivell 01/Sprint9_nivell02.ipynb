1. Comptador i endreçador de paraules d'un text.

El client va quedar content amb el comptador de paraules, però ara vol llegir arxius TXT i que calculi la freqüència de cada paraula ordenades dins de les entrades habituals del diccionari segons la lletra amb la qual comencen, és a dir, les claus han d'anar de la A a la Z i dins de la A hem d'anar de la A la Z. Per exemple, per a l'arxiu "tu_me_quieres_blanca.txt" la sortida esperada seria:

```python
import string

def comptar_paraules_per_inicial(fitxer):
    """
    Llegeix un fitxer .txt i retorna un diccionari amb les paraules
    agrupades per la seva inicial i amb la seva freqüència d’aparició.

    """
    # Llegim tot el text del fitxer
    with open(fitxer, 'r', encoding='utf-8') as f:
        text = f.read()

    # Netejar i dividir el text en paraules
    paraules = text.split()
    freq = {}

    for p in paraules:
        # Normalitzar: minúscules i eliminar signes de puntuació
        p = p.lower().strip(string.punctuation + "«»“”¿?¡!—…")
        if not p:
            continue  # Evitem paraules buides

        inicial = p[0].upper()  # primera lletra en majúscula

        if inicial.isalpha():  # només si és una lletra A-Z
            if inicial not in freq:
                freq[inicial] = {}
            freq[inicial][p] = freq[inicial].get(p, 0) + 1

    # Ordenar les claus del diccionari principal (A-Z)
    resultat = {}
    for lletra in sorted(freq.keys()):
        # Ordenar també les paraules dins de cada lletra
        subdicc_ordenat = dict(sorted(freq[lletra].items()))
        resultat[lletra] = subdicc_ordenat

    return resultat


# 🔹 Exemple d'ús
fitxer = "C:/LLUIS/Dropbox/IT ACADEMY/Sprint 9/Nivell 02/La Caputxeta vermella.txt"  # fitxer
resultat = comptar_paraules_per_inicial(fitxer)

# Mostrar el resultat d’una manera clara
for lletra, paraules in resultat.items():
    print(f"\n{lletra}:")
    for paraula, count in paraules.items():
        print(f"  {paraula}: {count}")
```

2. Conversió de tipus de dades.

El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client:

```python
conversions=([ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']])
llista_floats=[]
llista_no_floats=[]

for item in conversions:
    if isinstance(item,(list, tuple)):
        for subitem in item:
            try:
                llista_floats.append(float(subitem))
            except ValueError:
                llista_no_floats.append(subitem)
    else:
        try:
            llista_floats.append(float(item))
        except ValueError:
            llista_no_floats.append(item)
    
print("Llista de floats:", llista_floats)
print("Llista de no-floats:", llista_no_floats) 
```
