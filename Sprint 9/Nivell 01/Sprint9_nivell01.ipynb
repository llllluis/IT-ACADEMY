1. Calculadora de l'índex de massa corporal

- Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades. Pots obtenir més informació del seu càlcul en: 🔗 Índice de masa corporal IMC que es y como se calcula.

- La funció ha de classificar el resultat en les seves respectives categories.

Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduïdes per l’usuari/a no estan en el format adequat o no pren valors raonables.

```python
# CALCULAR MASS A CORPORAL - Sprint9_nivell 1.1
def calcular_imc(pes, altura):
    """
    Calcula el Índice de Masa Corporal (IMC).
    
    Parámetros:
        pes (float): peso en kilogramos
        altura (float): altura en metros
    
    Retorna:
        float: IMC calculat
    """
    altura = altura.replace(',', '.')
    altura = float(altura)
    pes=float(pes)
    if pes <= 0:
        raise ValueError("El pes ha de ser més gran que cero.")
    if altura <= 0:
        raise ValueError("La altura ha de ser més gran que cero.")   

    #càlcul del IMC
    imc = pes / (altura ** 2)

    if imc < 18.5:
        categoria=("Estàs per sota del pes normal.")
    elif 18.5 <= imc < 24.9:
        categoria=("Tens un pes normal.")
    elif 25 <= imc < 29.9:
        categoria=("Estàs en sobrepès.")
    else:
        categoria=("Estàs en obesitat.")
    
    #retona el IMC calculat
    return imc,categoria

# bucle que permet varios càlculs fins que l'usuari decideixi aturar-se
continuar = 's'
while continuar == 's': 
    try:
        pes = (input("Introdueix el teu pes en kg: "))
        altura = (input("Introdueix la teva alçada en metres: "))             
        imc, categoria = calcular_imc(pes, altura)
        print("----------------------------")
        print("Pes=", pes, "kg")
        print("Alçada=", altura, "m")   
        print(f"El teu Índex de Massa Corporal (IMC) és: {imc:.2f}")
        print(categoria)
        print("----------------------------")
        continuar = input("Vols calcular un altre IMC? (s/n): ").strip().lower()
    except ValueError as e:
        print(f"Error: {e}")
    except KeyboardInterrupt:
        print("\nCàlcul del IMC finalitzat.")
        break
```

2. Convertidor de temperatures

Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re).

```python
# Convertridor de temperatures - Sprint9_nivell 1.2
import pandas as pd
df = pd.read_excel(r"C:/LLUIS/Dropbox/IT ACADEMY/Sprint 9/Nivell 01/Tabla_conversio_temperatura.xlsx",index_col=0)
df.head()
Temperatures=[] 
def convertir_temperatura(valor, desde, hacia, df=df):
    """
    Convierte temperaturas entre Celsius, Fahrenheit y Kelvin.
    
    Parámetros:
        valor (float): temperatura a convertir
        desde (str): escala de origen ('C', 'F' o 'K')
        hacia (str): escala de destino ('C', 'F' o 'K')
    
    Retorna:
        float: temperatura convertida
    """
    desde, hacia = desde.upper(), hacia.upper()
    T=valor
    resultado= eval(df.loc[desde, hacia])
    #guardo les temperatures en una llista
    Temperatures.append({
        'ValorInicial': T,
        'Desde': desde,
        'hacia': hacia,
        'Resultat': resultado
    })
    return resultado

temperatura='s'
while temperatura=='s':
    try:
        temperatura = float(input("Ingrese la temperatura a convertir (o 'exit' para salir): "))
        desde = input("Ingrese la escala de origen (C, F, K): ")
        hacia = input("Ingrese la escala de destino (C, F, K): ")
        if desde.upper() not in ['C', 'F', 'K'] or hacia.upper() not in ['C', 'F', 'K']:
            print("Escala no válida. Por favor, use 'C', 'F' o 'K'.")
            continue            
        resultado = convertir_temperatura(temperatura, desde, hacia)        
        print("Historial de conversiones:")
        for temp in Temperatures:
            print(f"{temp['ValorInicial']} {temp['Desde']} = {temp['Resultat']:.2f} {temp['hacia']}")
        temperatura=input("¿Desea realizar otra conversión? (s/n): ").strip().lower()
    except ValueError:
        print("Entrada no válida. Por favor, intente de nuevo.")
    except KeyboardInterrupt:
        print("\nConversión terminada.")
        break
    
    

```

3. Comptador de paraules d'un text.

Escriu una funció que donat un text, mostri les vegades que apareix cada paraula. Intenta que es gestionin totes les casuístiques possibles que facin que el programa no funcioni correctament.

- (EXTRA): Quina és la llargada mitjana de les paraules del text que has escrit? “Hola com va?” hauria de retornar (4+3+2) / 3 = 3


```python
# Comptador de paraules d'un text. - Sprint9_nivell 1.3
# text per probar ="La programació fomenta el pensament lògic, la creativitat i la resolució de problemes. Aprendre a codificar des de jove obre oportunitats i desenvolupa habilitats essencials.""
def comptar_paraules(text):
    """
    Comptar la freqüència de cada paraula en un text.
    
    Paràmetres:
        text (str): text d'entrada
    
    Retorna:
        dict: diccionari amb la freqüència de cada paraula
    """
    paraules = text.split()
    freq = {}
    for paraula in paraules:
        paraula = paraula.lower().strip('.,!?;"()[]{}')  # Normalitzar la paraula
        if paraula:
            freq[paraula] = freq.get(paraula, 0) + 1
    return freq
text = input("Introdueix un text: ")
frequencia = comptar_paraules(text)
llargades=0
TotalParaules=0
print("Freqüència de paraules:")
for paraula, comptador in frequencia.items():
    print(f"{paraula}: {comptador}")
    # Calcular la suma total de les allargades de cada paraual i  mitjana d'alalrgada per paraula
    llargades += len(paraula) * comptador
    TotalParaules += (1*comptador)
print(f"Total de paraules: {TotalParaules}")
print(f"Suma de llargades de totes les paraules: {llargades}")    
mitjana = llargades / TotalParaules
print(f"Llargada mitjana de les paraules: {mitjana:.2f}")
    

```

4. Diccionari invers (amb possibilitat de duplicats)

```python
# Sprint9_nivell 1.4


reverse_dicc = ({'Vermell': 3, 'Blau': 5, 'Verd': 2, 'Groc': 4, 'Negre': 1})
reverse_dicc2=({'Gat':1,'Rana':5,'ocell':3,'peix':4,'Gos':5})

def reverse_dictionary(dicc):
    """
    Inverteix les claus i valors d'un diccionari.
    
    Paràmetres:
        dicc (dict): diccionari d'entrada
    
    Retorna:
        dict: diccionari amb claus i valors invertits
    """
    #return {valor: clau for clau, valor in dicc.items()}    
    inverted_dicc_retorn= {}
    for clau, valor in dicc.items():
        if valor in inverted_dicc_retorn: #Comprova si el valor actual (que volem convertir en nova clau) ja existeix dins del nou diccionari
            if isinstance(inverted_dicc_retorn[valor], list): #Comprova si el valor ja és una llista, vol dir que ja hi havia més d’una clau original amb el mateix valor.
                inverted_dicc_retorn[valor].append(clau) #Afegeix la nova clau a la llista existent.
            else: #Si el valor no és una llista, significa que és la primera vegada que es troba un valor duplicat.
                inverted_dicc_retorn[valor] = [inverted_dicc_retorn[valor], clau] #Crea una llista amb l'antiga clau i la nova clau.
        else: #Si el valor no existeix encara en el nou diccionari, simplement afegeix la nova clau i valor.
            inverted_dicc_retorn[valor] = clau #Afegeix la nova clau i valor al diccionari invertit.
            
    return inverted_dicc_retorn

print("Diccionari original 1:", reverse_dicc)
print("Diccionari original 2:", reverse_dicc2)  
print("Inversió de diccionaris:")   
print("Inversió de diccionari 1:", reverse_dictionary(reverse_dicc))
print("Inversió de diccionari 2:", reverse_dictionary(reverse_dicc2))   


```
