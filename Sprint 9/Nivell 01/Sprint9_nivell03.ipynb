1. Generador de Contrasenyes

Explora el funcionament del mòdul random de la llibreria numpy. 🔗 Random module in NumPy

En aquest punt, el client ha detectat un problema amb les contrasenyes que utilitzen els seus treballadors. Asdf1234, dates d’aniversari o similars. Per a solucionar-ho ens ha encarregat una funció de Python que generi contrasenyes més segures. La funció ha de dependre dels següents paràmetres:

- longitud (int): Longitud de la contrasenya
- majuscules (bool = True): Si hi ha d’aparèixer majúscules
- minuscules (bool = True): Si hi ha d’aparèixer minúscules
- numeros (bool = True): Si hi ha d’aparèixer números
- signes (bool = False): Si hi ha d’aparèixer caràcters especials (,-$? o similars)


```python
import random
import string  
def generar_contrasenya(longitud, majuscules=True, minuscules=True, numeros=True, signes=False):
    """
    Genera una contrasenya segura basada en els paràmetres donats.
    
    Args:
        longitud (int): Longitud de la contrasenya.
        majuscules (bool): Si hi ha d'aparèixer majúscules.
        minuscules (bool): Si hi ha d'aparèixer minúscules.
        numeros (bool): Si hi ha d'aparèixer números.
        signes (bool): Si hi ha d'aparèixer caràcters especials.
    
    Returns:
        str: La contrasenya generada.
    
    Raises:
        ValueError: Si no es selecciona cap tipus de caràcter o si la longitud és insuficient.
    """
    if not (majuscules or minuscules or numeros or signes):
        raise ValueError("Cal seleccionar almenys un tipus de caràcter.")
    
    tipus_caracters = ""
    if majuscules:
        tipus_caracters += string.ascii_uppercase
    if minuscules:
        tipus_caracters += string.ascii_lowercase
    if numeros:
        tipus_caracters += string.digits
    if signes:
        tipus_caracters += string.punctuation
    
    # --- print("Caracters per codificar = " + tipus_caracters)

    # definir variable contrasenya
    contrasenya = []    
    
    # assegurar que cada tipus de caràcter seleccionat apareix almenys una vegada
    if majuscules:
        contrasenya.append(random.choice(string.ascii_uppercase))
    if minuscules:
        contrasenya.append(random.choice(string.ascii_lowercase))
    if numeros:
        contrasenya.append(random.choice(string.digits))
    if signes:
        contrasenya.append(random.choice(string.punctuation))

    # completar la contrasenya fins a la longitud desitjada
    contrasenya += random.choices(tipus_caracters, k=longitud - len(contrasenya))
    random.shuffle(contrasenya)
    
    return ''.join(contrasenya)

input_longitud = input("Introdueix la longitud de la contrasenya (mínim 8 caràcters): ")
try:
    longitud = int(input_longitud)
    if longitud < 8:
        raise ValueError("La longitud mínima és de 8 caràcters.")
except ValueError as ve:
    print(f"Error: {ve}")
    exit(1) 
input_majuscules = input("Vols incloure majúscules? (s/n): ").strip().lower() == 's'
input_minuscules = input("Vols incloure minúscules? (s/n): ").strip().lower() == 's'
input_numeros = input("Vols incloure números? (s/n): ").strip().lower() == 's'
input_signes = input("Vols incloure caràcters especials? (s/n): ").strip().lower() == 's'
try:
    contrasenya = generar_contrasenya(longitud, majuscules=input_majuscules, minuscules=input_minuscules, numeros=input_numeros, signes=input_signes)
    print(f"\nContrasenya generada: {contrasenya}")
except ValueError as ve:
    print(f"Error: {ve}")
    exit(1) 
# creació de contrasenyes 
#print(f"\nContrasenya= " + generar_contrasenya(16, majuscules=True, minuscules=True, numeros=True, signes=True))
```

2. Processament de dades simple

Una companya de feina ens ha demanat un favor, aprofitant que sap que estem aprenent a programar. Té un històric de partits de futbol català en un fitxer, on hi ha emmagatzemat els noms dels equips i els resultats. Necessita que processem les dades de manera automàtica, per extreure’n els resultats que necessita.

Utilitza el fitxer "historic_partits.txt"

Necessita un programa que retorni:

- El nombre total de gols que ha fet cada equip.
- El nom de l’equip més golejador.
- El nom de l’equip més golejat
- La classificació global (cada victòria: 3 pts, empat 1 pts, derrota 0 pts)

```python

import pandas as pd

# --- Llegim el fitxer i extraiem la informació dels partits ---
partits = []
fitxer = "C:/LLUIS/Dropbox/IT ACADEMY/Sprint 9/Nivell 03/historic partits.txt"
with open(fitxer, "r", encoding="utf-8") as f:
    for linia in f:
        linia = linia.strip()
        if not linia:
            continue
        # Exemple de línia: "Vilafranca\t4-5\tTerrassa"
        parts = linia.split("\t")
        if len(parts) == 3:
            equip_local = parts[0].strip()
            marcador = parts[1].strip()
            equip_visitant = parts[2].strip()

            # Separem els gols
            gols_local, gols_visitant = map(int, marcador.split("-"))

            # Afegim el partit a la llista
            partits.append((equip_local, gols_local, equip_visitant, gols_visitant))

# ---  Crear la llista d'equips únics ---
    #  Obtenir equips locals
    locals = [p[0] for p in partits]

    #  Obtenir equips visitants
    visitants = [p[2] for p in partits]

    # Unir totes les llistes
    tots = locals + visitants

    # Eliminar duplicats
    sense_repetits = set(tots)

    # Ordenar alfabèticament
    equips = sorted(sense_repetits)

    #equips = sorted(set([p[0] for p in partits] + [p[2] for p in partits]))

# --- Inicialitzar el diccionari d'estadístiques per a cada equip ---
#  Creem un diccionari buit per guardar les estadístiques
stats = {}

# ecorrem la llista d’equips
for e in equips:
    
    # Creem un diccionari amb les estadístiques inicials d’aquest equip
    stats[e] = {
        'Partits Jugats': 0,
        'Victòries': 0,
        'Empats': 0,
        'Derrotes': 0,
        'Gols a Favor': 0,
        'Gols en Contra': 0,
        'Diferència de Gols': 0,
        'Punts': 0
    }

#  "stats" és un diccionari on:
#  - cada clau és el nom d’un equip
#  - cada valor és un altre diccionari amb totes les seves estadístiques


# --- Processar cada partit i actualitzar estadístiques ---
for local, gl, visit, gv in partits:
    # Partits jugats
    stats[local]['Partits Jugats'] += 1
    stats[visit]['Partits Jugats'] += 1

    # Gols a favor i en contra
    stats[local]['Gols a Favor'] += gl
    stats[local]['Gols en Contra'] += gv
    stats[visit]['Gols a Favor'] += gv
    stats[visit]['Gols en Contra'] += gl

    # Resultat del partit
    if gl > gv:
        # victòria local
        stats[local]['Victòries'] += 1
        stats[visit]['Derrotes'] += 1
        stats[local]['Punts'] += 3
    elif gl < gv:
        # victòria visitant
        stats[visit]['Victòries'] += 1
        stats[local]['Derrotes'] += 1
        stats[visit]['Punts'] += 3
    else:
        # empat
        stats[local]['Empats'] += 1
        stats[visit]['Empats'] += 1
        stats[local]['Punts'] += 1
        stats[visit]['Punts'] += 1

# --- Calcular la diferència de gols ---
for e in equips:
    stats[e]['Diferència de Gols'] = stats[e]['Gols a Favor'] - stats[e]['Gols en Contra']

# --- Crear un DataFrame amb els resultats ---
df_stats = pd.DataFrame.from_dict(stats, orient='index')

# Ordenar per punts (descendent)
df_stats = df_stats.sort_values(by=['Punts', 'Diferència de Gols'], ascending=False)



# --- Mostrar l'equip amb mes gols a favor ---
equip_mes_gols = df_stats['Gols a Favor'].idxmax()
gols_mes_alts = df_stats['Gols a Favor'].max()
print(f"\nL'equip amb més gols a favor és {equip_mes_gols} amb {gols_mes_alts} gols.")

# --- Mostrar l'equip amb mes gols en contra ---
equip_mes_contra = df_stats['Gols en Contra'].idxmax()
gols_mes_contra = df_stats['Gols en Contra'].max()
print(f"L'equip amb més gols en contra és {equip_mes_contra} amb {gols_mes_contra} gols.")

# --- Mostrar la classificació final ---
df_stats

```
