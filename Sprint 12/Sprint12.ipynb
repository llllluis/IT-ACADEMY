## 4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.

- Mostra la resposta JSON.
- Mostra el codi d'estat.

```python
import requests
dades = {'title': 'Titol post01 luigi', 'body': 'luigi Cos post 01', 'UserID': 1}
url='https://jsonplaceholder.typicode.com/posts'
r = requests.post(url, data=dades)
print(r.text)
print(f"Estat petició:{r.status_code}")
```

## 5. Fes una petició PATCH per modificar parcialment una publicació existent.

- Mostra la resposta JSON.
- Mostra el codi d'estat.

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts/1"

# Només modifiquem el camp que ens interessa (exemple: el títol)
payload = {
    "title": "Títol actualitzat parcialment"
}

response = requests.patch(url, json=payload)


print(response.text)
print("Codi d'estat:", response.status_code)
```

### Selecciona --> API de gestió del catàleg --> https://opendata-ajuntament.barcelona.cat/data/api/action/

// Aquests mètodes són públics i no requereixen cap tipus d'identificació ni token sent alguns exemples:

    current_package_list_with_resources: retorna el llistat de datasets actuals i els seus recursos
    package_search: fa una cerca dins del catàleg de datasets
    resource_show: retorna les metadades del dataset a partir de l'id del dataset i entre elles el resource_id dels recursos
- Exemple: https://opendata-ajuntament.barcelona.cat/data/api/3/action/package_show?id=arbrat-zona
datastore_search: retorna la informació de cada recurs.
- Exemple: https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search?resource_id=f7134b6f-95ce-4b83-bacd-6f411bd9c1f0
i si a més volguéssim buscar, per exemple, el valor=14 dins el recurs:
https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search?resource_id=f7134b6f-95ce-4b83-bacd-6f411bd9c1f0&q=14

### 1. Utilitza l’API d’Open Data BCN per:

Cercar un dataset del teu interès mitjançant package_search i package_show. .


```python
import requests
import pandas as pd

# 1) Buscar un dataset
resp = requests.get(
    "https://opendata-ajuntament.barcelona.cat/data/api/3/action/package_search",
    params={"q": "educacio"}
)
data = resp.json()

# fem un llistat
#for result in data["result"]["results"]:
#    print(f"- {result['title']} (ID: {result['name']}) ({result['num_resources']} recursos)({result['organization']['title']})")    

for i, result in enumerate(data["result"]["results"], start=1):
    print(f"{i}. - {result['title']} (ID: {result['name']}) "
          f"({result['num_resources']} recursos) ({result['organization']['title']})")

input_idx = input("Tria un dataset (1-{}): ".format(len(data["result"]["results"])))
idx = int(input_idx) - 1
# agafo el priemr de la llista
titulo = data["result"]["results"][idx]["title"]    
print(f"Opció triada ({input_idx}):", titulo)
dataset_id = data["result"]["results"][idx]["name"]    
print(f"Dataset triat: ", dataset_id)



# 2) Obtenir metadades del dataset
resp_meta = requests.get(
    "https://opendata-ajuntament.barcelona.cat/data/api/3/action/package_show",
    params={"id": dataset_id}
)
meta = resp_meta.json()

resource_id = meta["result"]["resources"][0]["id"]
print("Resource ID: ", resource_id)


```

### 2. Dels resultats obtinguts, selecciona’n un que tingui recursos disponibles en CSV o JSON.

```python

# agafo el  recurs amb format json.
"""
- 2025_pad_mdbas_niv-educa-esta_sexe.csv
  format: CSV
  resource_id: 48ed4d39-78a1-49ee-9f3f-ba4872b678b6
"""
print("----- DADES RECURS TRIAT -----")
print(f"id Recurs triat : {resource_id}")
print("------------------------------")
#resource_id = "48ed4d39-78a1-49ee-9f3f-ba4872b678b6"
```

### 4. Converteix els resultats a un DataFrame.

```python
import requests
import pandas as pd

#resource_id = "48ed4d39-78a1-49ee-9f3f-ba4872b678b6"
  

# Endpoint datastore_search
url = "https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search"

params = {
    "resource_id": resource_id,
    "limit": 100  # com a mínim 100 registres
}

response = requests.get(url, params=params)
response.raise_for_status()

data = response.json()

if not data.get("success"):
    raise Exception("Error en la consulta API")

# Convertir registres a DataFrame
records = data["result"]["records"]
df = pd.DataFrame(records)

print("Total registres recuperats:", len(df))
df.head()


```

### 5. Desa el DataFrame en un fitxer .csv.

```python
# 4) Desa el DataFrame a CSV
df.to_csv("dataset_educacio.csv", index=False, encoding="utf-8")

print("Fitxer desat com: dataset_educacio.csv")
```
