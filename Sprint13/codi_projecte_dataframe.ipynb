## Carrego el dataframe de cursos escolars del fitxer escel -> tbEJERCICIO_ESCOLAR.xlsx

```python
import pandas as pd
## carrego el fitxer excell tbEJERCICIO_ESCOLAR.xlsx
df_cursos_escolars = pd.read_excel('C:\\temp\\tbEJERCICIO_ESCOLAR.xlsx')
df_cursos_escolars.head(100)
```

## Creo un camp nou al dataframe principal que es __INICI_CURS__  mapenjant-lo del dataframe df_cursos_escolars


```python
# creo el camp INICI al dataframe principal  mapenjant-los del df_cursos_escolars
df_principal['INICI_CURS'] = df_principal['NRO_EJERCICIO'].map(df_cursos_escolars.set_index('NRO_EJERCICIO')['INICI'])
df_principal.head(5)
```

## Creo un camp nou al dataframe principal que es __EDAD__  
### El calcularem a partir del camp INICI del  camp DATANAIXEMENT del df_principal per saber els anys que tenia el alumne quan va fer el curs.

```python
# creo camp nou al dataframe principal que es EDAD a partir del camp INICI del  camp DATANAIXEMENT del df_principal per saber els anys que tenia el alumne quan va fer el curs.  
from datetime import date, datetime

def calcular_edat_simplificada(data_naixement, data_referencia):
    """Versión simplificada para calcular edad entre dos fechas"""
    try:
        # Convertir a datetime si es necesario
        dn = pd.to_datetime(data_naixement)
        dr = pd.to_datetime(data_referencia)
        
        # Calcular edad exacta
        edad = dr.year - dn.year - ((dr.month, dr.day) < (dn.month, dn.day))
        return int(round(edad))
    except:
        return None

# Aplicar la función
df_principal['EDAD'] = df_principal.apply(
    lambda row: calcular_edat_simplificada(row['DATANAIXEMENT'], row['INICI_CURS']),
    axis=1
).astype('Int64')  # Usar Int64 para permitir NaN

# Ver resultados
print(df_principal[['DATANAIXEMENT', 'INICI_CURS', 'EDAD']].head())
print(f"Valores NaN en EDAD: {df_principal['EDAD'].isna().sum()}")

```
