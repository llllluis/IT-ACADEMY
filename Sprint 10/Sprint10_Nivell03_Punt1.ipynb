```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os

# ---------------------------
# FUNCIONS DE GRÀFIC
# ---------------------------

def grafic_numeric(serie, nom_columna, carpeta,tipus="histograma"):
    """Crea i guarda un histograma i/o boxplot per a una columna numèrica."""

    if tipus == "histograma":
        plt.figure(figsize=(6,4))
        sns.histplot(serie, kde=True)
        plt.title(f"Histograma de {nom_columna}")
        plt.tight_layout()
        plt.savefig(f"{carpeta}/{nom_columna}_histograma.png")
        plt.close()
            
    elif tipus == "boxplot":
        plt.figure(figsize=(6,4))
        sns.boxplot(x=serie)
        plt.title(f"Boxplot de {nom_columna}")
        plt.tight_layout()
        plt.savefig(f"{carpeta}/{nom_columna}_boxplot.png")
        plt.close()
    else: # ambdós    
        plt.figure(figsize=(10,4))
        
        # Histograma
        plt.subplot(1,2,1)
        sns.histplot(serie, kde=True)
        plt.title(f"Histograma de {nom_columna}")
        
        # Boxplot
        plt.subplot(1,2,2)
        sns.boxplot(x=serie)
        plt.title(f"Boxplot de {nom_columna}")
    
        plt.tight_layout()
        plt.savefig(f"{carpeta}/{nom_columna}_numeric.png")
        plt.close()


def grafic_categoric(serie, nom_columna, carpeta):
    """Crea i guarda un gràfic de barres per a una columna categòrica."""
    plt.figure(figsize=(8,5))
    valors = serie.value_counts().head(10)
    sns.barplot(x=valors.values, y=valors.index)
    plt.title(f"Valors més freqüents de {nom_columna}")
    plt.xlabel("Comptatge")
    plt.ylabel(nom_columna)
    plt.tight_layout()
    plt.savefig(f"{carpeta}/{nom_columna}_categoric.png")
    plt.close()


def grafic_data(serie, nom_columna, carpeta):
    """Crea i guarda un gràfic de barres amb la distribució d'anys d'una data."""
    plt.figure(figsize=(8,5))
    anys = serie.dt.year.value_counts().sort_index()
    sns.barplot(x=anys.index, y=anys.values)
    plt.title(f"Distribució per anys de {nom_columna}")
    plt.xlabel("Any")
    plt.ylabel("Comptatge")
    plt.tight_layout()
    plt.savefig(f"{carpeta}/{nom_columna}_data.png")
    plt.close()


# ---------------------------
# FUNCIÓ PRINCIPAL
# ---------------------------

def graficar_dataframe(df, carpeta_sortida="C:\\LLUIS\Dropbox\\IT ACADEMY\\Sprint 10\\graficos_dataframe"):
    """
    Detecta el tipus de cada columna i crida la funció corresponent.
    """
    os.makedirs(carpeta_sortida, exist_ok=True)
    
    for columna in df.columns:
        dades = df[columna].dropna() # borro els valors nuls
        print(f"Procesant columna: {columna}...")
        
        # Detectem el tipus
        if pd.api.types.is_datetime64_any_dtype(dades):
            tipus = "data"
        elif pd.api.types.is_numeric_dtype(dades):
            tipus = "numeric"
        else:
            tipus = "categoric"
        
        # Cridem la funció corresponent
        if tipus == "numeric":
            grafic_numeric(dades, columna, carpeta_sortida, tipus="els 2")  # pots canviar a "boxplot" o "ambdós"
        elif tipus == "categoric":
            grafic_categoric(dades, columna, carpeta_sortida)
        elif tipus == "data":
            grafic_data(dades, columna, carpeta_sortida)
        else:
            print(f"No s'ha pogut identificar el tipus de la columna {columna}")
    
    print(f"Gràfics guardats a la carpeta: '{carpeta_sortida}'")

import seaborn as sns

# Carreguem un dataset d'exemple
df = sns.load_dataset("penguins")

# Cridem la funció
graficar_dataframe(df)
```
