- Crea una columna nova que sigui la data de naixament en format Datetime a partir de les columnes dia, mes i any. Utilitzant aquesta columna crea una funció que donada una data, et calculi l'edat actual a dia d'avui.

- Utilitza la funció que acabes de crear per generar una columna nova al DataFrame amb l'edat actual.

```python
from datetime import date, datetime

def calcular_edat(data_naixement):
    """
    Rep una data de naixement (datetime.date o cadena) i retorna l'edat en anys.
    """
    # Si la data ve com a cadena, la convertim a objecte date
    if isinstance(data_naixement, str):
        try:
            data_naixement = datetime.strptime(data_naixement, "%Y-%m-%d").date()
        except ValueError:
            data_naixement = datetime.strptime(data_naixement, "%d/%m/%Y").date()

    avui = date.today()
    edat = avui.year - data_naixement.year - ((avui.month, avui.day) < (data_naixement.month, data_naixement.day))
    return edat


# Crear un diccionari amb les parts de la data
dates = {'year': df['Any'], 'month': df['Mes'], 'day': df['Dia']}

# Convertir a dates vàlides (els errors es converteixen a NaT)
df['naixament'] = pd.to_datetime(dates, errors='coerce')

# Calcular l'edat i crear la nova columna 'Edat'
df['Edat'] = df['naixament'].apply(calcular_edat)
df.head(10)
```

- Actualitza la columna salari en funció dels percentatges que s'adjunten. No modifiquis manualment els increments, escriu codi Python per fer les conversions necessàries.

```python
def calcular_increment(salari, increment):
    """
    Rep un salari (float) i un increment (cadena amb percentatge, ex. '5%' o '3,5%') i retorna el nou salari amb l'increment aplicat.
    """
    if pd.isna(salari) or pd.isna(increment):
        return salari  # Si algun valor és NaN, retornem el salari original

    # Convertim l'increment a float
    increment = increment.replace('%', '').replace(',', '.').strip()
    try:
        increment_float = float(increment) / 100.0
    except ValueError:
        return salari  # Si no es pot convertir, retornem el salari original

    nou_salari = salari * (1 + increment_float)
    return round(nou_salari, 2)

df_increment = pd.DataFrame({"Grup":["Grup A","Grup B","Grup C", "Grup D" ] , "Increment":["5%","3,5%","2%","8%"]})

df['increment'] = df['Grup Professional'].map(df_increment.set_index('Grup')['Increment'])
df['Salari actualitzat'] = df.apply(lambda row: calcular_increment(row['Salari numeric'], row['increment']), axis=1)
df[['Nom_Complert', 'Grup Professional', 'Salari mensual', 'Salari numeric', 'increment', 'Salari actualitzat']].head(10)
```

Utilitzant un bucle, exporta en 4 fitxers (format .xlsx o .csv) les dades de cada Grup Professional.

Per exemple: "dades_GrupA.xlsx" , "dades_GrupB.xlsx" ...

Exporta un 5è DataFrame en format .xlsx o .csv que contingui quants treballadors hi ha per cada Grup Professional, quin és el seu sou mig i quina és la seva edat mediana.

```python
for grup in df['Grup Professional'].unique():
    df_grup = df[df['Grup Professional'] == grup]
    nom_fitxer = f'dades_{grup.replace(" ", "")}.csv'  # o .xlsx si prefereixes aquest format
    df_grup.to_csv(nom_fitxer, index=False)
    df_grup.to_excel(nom_fitxer.replace('.csv', '.xlsx'), index=False)
    print(f'Fitxer creat: {nom_fitxer}')

# Exporta un 5è DataFrame en format .xlsx o .csv que contingui quants treballadors hi ha per cada Grup Professional, 
# quin és el seu sou mig i quina és la seva edat mediana.
df_resum_grup = df.groupby('Grup Professional').agg(
    Num_Treballadors=('Nom', 'count'),
    Sou_Mig=('Salari numeric', 'mean'),
    Edat_Mediana=('Edat', 'median')
).round(2)
df_resum_grup.to_csv('resum_grup_professional.csv')
df_resum_grup.to_excel('resum_grup_professional.xlsx')
df_resum_grup
```
